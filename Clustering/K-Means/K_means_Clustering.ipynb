```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset
```

```python
x=dataset.iloc[:,3:5].values
```

```python
x
```

```python
list1=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=36)
    kmeans.fit(x)
    list1.append(kmeans.inertia_)
plt.plot(range(1,11),list1)
plt.title("Elbow Method")
plt.xlable("Number_of_clusters")
plt.ylable("distance")
```

```python
kmeans=KMeans(n_clusters=5,init='k-means++',random_state=36)
kmeans.fit(x)
```

```python
y_pred=kmeans.predict(x)
```

```python
y_pred
```

```python
dataset["Cluster_values"]=y_pred
dataset
```

```python
dataset.to_csv("cluster.csv",index=False)
```

```python
centroids=kmeans.cluster_centers_
centroids
```
