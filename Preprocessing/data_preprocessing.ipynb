```python
quan=[] #numerical
qual=[] #categorical

for columnNames in dataset.columns:
    if(dataset[columnNames].dtype =="str"):
        qual.append(columnNames)
    else:
        quan.append(columnNames)
```

```python
descriptive = pd.DataFrame(index = ["Mean", "Median", "Mode", "Min", "25%", "50%", "75%", "Max","IQR","1.5IQR","LesserOutlier","GreaterOutlier"], columns=quan)
for columnName in quan:
    descriptive.loc["Mean", columnName] = dataset[columnName].mean()
    descriptive.loc["Median",columnName] = dataset[columnName].median()
    descriptive.loc["Mode",columnName] = dataset[columnName].mode()[0]
    descriptive.loc["Min",columnName] = dataset.describe()[columnName]["min"]
    descriptive.loc["25%",columnName] = dataset.describe()[columnName]["25%"]
    descriptive.loc["50%",columnName] = dataset.describe()[columnName]["50%"]
    descriptive.loc["75%",columnName] = dataset.describe()[columnName]["75%"]
    descriptive.loc["Max",columnName] = dataset.describe()[columnName]["max"]
    descriptive.loc["IQR",columnName] = dataset.describe()[columnName]["75%"] - dataset.describe()[columnName]["25%"]
    descriptive.loc["1.5IQR",columnName] =  1.5 * descriptive[columnName]["IQR"]
    descriptive.loc["LesserOutlier",columnName] = dataset.describe()[columnName]["25%"] - descriptive[columnName]["1.5IQR"]
    descriptive.loc["GreaterOutlier",columnName] = dataset.describe()[columnName]["75%"] + descriptive[columnName]["1.5IQR"]
```

```python
descriptive
```

```python
lesser=[]
greater=[]
for columnNames in quan:
    if descriptive[columnNames]["LesserOutlier"] > descriptive[columnNames]["Min"]:
        lesser.append(columnNames)
    if descriptive[columnNames]["GreaterOutlier"] < descriptive[columnNames]["Max"]:
        greater.append(columnNames)
```

```python
greater
```

```python

for lesscolumn in lesser:
    dataset.loc[dataset[lesscolumn] < descriptive.loc["LesserOutlier", lesscolumn], lesscolumn] = descriptive.loc["LesserOutlier", lesscolumn]

for greatcolumn in greater:
    dataset.loc[dataset[greatcolumn] > descriptive.loc["GreaterOutlier", greatcolumn], greatcolumn] = descriptive.loc["GreaterOutlier", greatcolumn]
```
