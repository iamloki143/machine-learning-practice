```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 150, 120, 180, 200],
    "Profit": [20, 30, 25, 40, 45],
    "Marketing": [10, 20, 15, 30, 35],
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor", "Printer"],
    "Quantity": [40, 60, 30, 25, 20]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Bar Chart
plt.bar(df["Product"], df["Sales"])
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
#Line Graph
plt.plot(df["Month"], df["Sales"], marker="d")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Pie Chart
plt.pie(df["Sales"],labels=df["Product"],autopct="%1.1f%%")
plt.title("Sales Percentage")
plt.show()
```

```python
#Histogram
plt.hist(df["Sales"], bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatter Plot
plt.scatter(df["Marketing"], df["Sales"])
plt.title("Marketing vs Sales")
plt.xlabel("Marketing")
plt.ylabel("Sales")
plt.show()
```

```python
#Box Plot
plt.boxplot(df["Sales"])
plt.title("Sales Box Plot")
plt.show()
```

```python
#Horizontal Bar Chart
plt.barh(df["Product"], df["Sales"])
plt.title("Sales by Product")
plt.xlabel("Sales")
plt.show()
```

```python
#Fill Between
plt.fill_between(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Stack Plot
plt.stackplot(df["Month"],
              df["Sales"],
              df["Profit"],
              labels=["Sales", "Profit"])
plt.legend()
plt.title("Sales and Profit")
plt.show()
```

```python
#Heat Map
data = [
    df["Sales"],
    df["Profit"],
    df["Marketing"]
]
plt.imshow(data)
plt.xticks(range(5), df["Month"])
plt.yticks(range(3), ["Sales", "Profit", "Marketing"])
plt.colorbar()
plt.title("Heatmap")
plt.show()
```
