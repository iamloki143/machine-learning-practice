```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
dataset = dataset.drop("User ID", axis = 1)
```

```python
dataset
```

```python
x=dataset[['Gender_Male','Age','EstimatedSalary']]
y=dataset['Purchased']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=24)
```

```python
# Gaussian Naive Bayes
from sklearn.naive_bayes import GaussianNB
GNB = GaussianNB()
GNB.fit(x_train ,y_train)
y_pred=GNB.predict(x_test)
GNB_cm = confusion_matrix(y_test , y_pred)
GNB_clf_report = classification_report(y_test,y_pred)
```

```python
GNB_cm
```

```python
print(GNB_clf_report)
```

```python
# Multinomial Naive Bayes
from sklearn.naive_bayes import MultinomialNB
MNB = MultinomialNB(alpha=1.0, force_alpha=True, fit_prior=True, class_prior=None)
MNB.fit(x_train , y_train)
y_pred=MNB.predict(x_test)
MNB_cm = confusion_matrix(y_test , y_pred)
MNB_clf_report = classification_report(y_test,y_pred)
```

```python
MNB_cm
```

```python
print(MNB_clf_report)
```

```python
# Complement Naive Bayes
from sklearn.naive_bayes import ComplementNB
CNB= ComplementNB(alpha=1.0, force_alpha=True, fit_prior=True, class_prior=None, norm=False)
CNB.fit(x_train , y_train)
y_pred=CNB.predict(x_test)
CNB_cm = confusion_matrix(y_test , y_pred)
CNB_clf_report = classification_report(y_test,y_pred)
```

```python
CNB_cm 
```

```python
print(CNB_clf_report)
```

```python
# Bernoulli Naive Bayes
from sklearn.naive_bayes import BernoulliNB 
BNB= BernoulliNB(alpha=1.0, force_alpha=True, fit_prior=True, class_prior=None)
BNB.fit(x_train , y_train)
y_pred=BNB.predict(x_test)
BNB_cm = confusion_matrix(y_test , y_pred)
BNB_clf_report = classification_report(y_test,y_pred)
```

```python
BNB_cm
```

```python
print(BNB_clf_report)
```

```python
# Categorical Naive Bayes
from sklearn.naive_bayes import CategoricalNB 
CaNB= CategoricalNB (alpha=1.0, force_alpha=True, fit_prior=True, class_prior=None)
CaNB.fit(x_train , y_train)
y_pred=CaNB.predict(x_test)
CaNB_cm = confusion_matrix(y_test , y_pred)
CaNB_clf_report = classification_report(y_test,y_pred)
```

```python
print(CaNB_clf_report)
```
