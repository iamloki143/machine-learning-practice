```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
```

```python
dataset = pd.read_csv("50_Startups_dataset.csv")
dataset
```

```python
dataset.isnull().sum()
```

```python
x=dataset[['R&D Spend','Administration','Marketing Spend']]
y=dataset['Profit']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.20 , random_state=24)
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
grid_para = {'random_state':[24,36,42],'n_estimators':[100,200,250,300],'loss':['linear','square','exponential']}
```

```python
grid =GridSearchCV( AdaBoostRegressor(),grid_para,refit=True,verbose=3,n_jobs=-1)
grid.fit(x_train,y_train)
```

```python
y_pred=grid.predict(x_test)
y_pred
```

```python
score=r2_score(y_test,y_pred)
score
```

```python
re=grid.cv_results_ 
print("TheR_scorevalueforbestparameter{}:".format(grid.best_params_))
```
