```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
```

```python
dataset = pd.read_csv("insurance.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
x=dataset[['age','bmi','children','sex_male','smoker_yes','region_northwest','region_southeast','region_southwest']]
y=dataset['charges']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=24)
```

```python
x_train
```

```python
y_train
```

```python
sc = StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
para_grid = {'criterion':['squared_error','absolute_error','poisson'], 'max_features': ['auto','sqrt','log2'], 'n_estimators':[10,100]}
```

```python
grid = GridSearchCV(RandomForestRegressor(),para_grid, refit=True , verbose = 3 ,n_jobs = -1)
grid.fit(x_train , y_train)
```

```python

re=grid.cv_results_ 
print("TheR_scorevalueforbestparameter{}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
y_pred=grid.predict(x_test)
y_pred
```

```python
score = r2_score(y_test,y_pred)
score
```
